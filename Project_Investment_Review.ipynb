```python
import pandas as pd

import matplotlib.pyplot as plt


# بيانات اختبارية لمحفظة الاستثمارات (المبالغ بملايين الريالات، العوائد بالنسبة المئوية)
df = pd.DataFrame({

    "Investment": ["Digital Platform", "Smart Logistics", "Customer App",
                   "Cloud Migration", "Data Hub", "IoT Sensors"],

    "Capital_SAR_M": [4.0, 3.5, 2.0, 5.0, 2.5, 1.5],

    "Expected_ROI": [12, 10, 9, 11, 10, 8],

    "Actual_ROI": [14, 3, 4, 12, 9, -2]

})


hurdle_rate = 8   # الحد الأدنى المقبول للعائد
max_shortfall = 5  # أقصى فجوة مقبولة عن العائد المتوقع (نقاط مئوية)


print(df)
```

```python
# الرد على الاعتراض المتوقع: تعريف "ضعيف الأداء" ومعايير إعادة التقييم
# المعيار 1: العائد الفعلي أقل من الحد الأدنى المقبول (8%)
# المعيار 2: العائد الفعلي أقل من المتوقع بـ 5 نقاط مئوية أو أكثر
# الاستثمار الذي يحقق المعيارين معاً => يُحال لإعادة التقييم
# الاستثمار الذي يحقق معياراً واحداً فقط => يُوضع تحت المراقبة

df["Shortfall"] = df["Expected_ROI"] - df["Actual_ROI"]

df["Below_Hurdle"] = df["Actual_ROI"] < hurdle_rate

df["Large_Shortfall"] = df["Shortfall"] >= max_shortfall


def classify(row):

    if row["Below_Hurdle"] and row["Large_Shortfall"]:

        return "Re-evaluate"

    if row["Below_Hurdle"] or row["Shortfall"] > 0:

        return "Monitor"

    return "On Track"


df["Status"] = df.apply(classify, axis=1)


print(df[["Investment", "Actual_ROI", "Shortfall", "Below_Hurdle", "Large_Shortfall", "Status"]].to_string(index=False))


at_risk = df.loc[df["Status"] == "Re-evaluate", "Capital_SAR_M"].sum()

share = at_risk / df["Capital_SAR_M"].sum() * 100

print(f"\nCapital in investments to re-evaluate: SAR {at_risk:.1f}M ({share:.0f}% of portfolio)")


plt.figure(figsize=(11, 5))


plt.bar(df["Investment"], df["Shortfall"])


plt.axhline(

    max_shortfall,

    linestyle="--",

    label="Re-evaluation threshold = 5 pts"

)


plt.title("IoT Sensors Has the Largest Shortfall vs. Expected ROI")

plt.xlabel("Investment")

plt.ylabel("Shortfall vs. Expected ROI (percentage points)")

plt.legend()


plt.show()
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# ==========================================================
# Task 1 & 2 — Framing & Big Idea Construction
# ==========================================================
audience = "Investment Committee"
decision = "Which underperforming investments should be re-evaluated?"
objection = "How do we define 'underperforming' and what are the criteria for re-evaluation?"

# بناء العناصر الثلاثة للـ Big Idea
POV = "Three of six investments (SAR 7.0M, 38% of portfolio capital) are returning below the 8% hurdle rate."
STAKES = "Continuing to fund them ties up capital that could earn returns elsewhere and slows overall portfolio progress."
ACTION = "Approve a 90-day re-evaluation of these three investments, ending in a continue / restructure / exit decision for each."

big_idea = f"{POV} {STAKES} {ACTION}"

print("==================================================")
print("AUDIENCE:", audience)
print("DECISION:", decision)
print("ANTICIPATED OBJECTION:", objection)
print("--------------------------------------------------")
print("BIG IDEA:")
print(big_idea)
print("--------------------------------------------------")
print("Action:")
print(ACTION)
print("==================================================\n")

# ==========================================================
# Task 3 — Narrative Arc & Audit
# ==========================================================
storyboard = [
    ["BLUF", "Re-evaluate the three investments that fail both performance criteria."],
    ["Situation", "Returns vary widely across the six active investments."],
    ["Complication", "Three investments return less than the 8% hurdle rate."],
    ["Evidence", "The same three also fall 5+ points short of their expected ROI."],
    ["Options", "Keep funding all investments, or re-evaluate the failing ones."],
    ["Recommendation", "Launch a 90-day re-evaluation of the three failing investments."],
    ["Ask", "Approve the re-evaluation and pause new funding until it concludes."]
]

def audit(storyboard):
    checks = {
        "Exactly 7 scenes": len(storyboard) == 7,
        "Starts with BLUF": storyboard[0][0] == "BLUF",
        "Ends with Ask": storyboard[-1][0] == "Ask",
        "Every scene has a message": all(msg.strip() for _, msg in storyboard)
    }
    print("--- STORYBOARD AUDIT ---")
    for check, result in checks.items():
        print(f"{'PASS' if result else 'FAIL'} — {check}")
    return all(checks.values())

audit(storyboard)

# ==========================================================
# Task 4 — Render Scenes (Investment Portfolio Data)
# ==========================================================
# بيانات اختبارية لمحفظة الاستثمارات
df = pd.DataFrame({
    "Investment": ["Digital Platform", "Smart Logistics", "Customer App",
                   "Cloud Migration", "Data Hub", "IoT Sensors"],
    "Capital_SAR_M": [4.0, 3.5, 2.0, 5.0, 2.5, 1.5],   # رأس المال بملايين الريالات
    "Expected_ROI": [12, 10, 9, 11, 10, 8],            # العائد المتوقع %
    "Actual_ROI": [14, 3, 4, 12, 9, -2]                # العائد الفعلي %
})

hurdle_rate = 8     # الحد الأدنى المقبول للعائد
max_shortfall = 5   # أقصى فجوة مقبولة عن العائد المتوقع (نقاط مئوية)

# معايير إعادة التقييم (الرد على الاعتراض المتوقع)
df["Shortfall"] = df["Expected_ROI"] - df["Actual_ROI"]
df["Status"] = [
    "Re-evaluate" if (a < hurdle_rate and s >= max_shortfall)
    else "Monitor" if (a < hurdle_rate or s > 0)
    else "On Track"
    for a, s in zip(df["Actual_ROI"], df["Shortfall"])
]

status_colors = {"Re-evaluate": "#dc2626", "Monitor": "#f59e0b", "On Track": "#cbd5e1"}

os.makedirs("investment_review_solution", exist_ok=True)

# Scene 1 — Situation
plt.figure(figsize=(11, 4.5))
plt.bar(df["Investment"], df["Actual_ROI"], color="#cbd5e1")
plt.axhline(0, color="gray", linewidth=0.8)
plt.title("Returns Vary Widely Across Active Investments", fontweight="bold")
plt.xlabel("Investment")
plt.ylabel("Actual ROI (%)")
plt.ylim(-5, 16)
plt.savefig("investment_review_solution/scene1_situation.png", bbox_inches="tight")
plt.show()

# Scene 2 — Complication
plt.figure(figsize=(11, 4.5))
plt.bar(df["Investment"], df["Actual_ROI"],
        color=["#cbd5e1" if x >= hurdle_rate else "#dc2626" for x in df["Actual_ROI"]])
plt.axhline(hurdle_rate, linestyle="--", color="black", label="Hurdle Rate = 8%")
plt.axhline(0, color="gray", linewidth=0.8)
plt.title("Three Investments Return Below the 8% Hurdle Rate", fontweight="bold", color="#b91c1c")
plt.xlabel("Investment")
plt.ylabel("Actual ROI (%)")
plt.ylim(-5, 16)
plt.legend()
plt.savefig("investment_review_solution/scene2_complication.png", bbox_inches="tight")
plt.show()

# Scene 3 — Evidence (Shortfall vs. Expected ROI)
plt.figure(figsize=(11, 4.5))
plt.bar(df["Investment"], df["Shortfall"], color=[status_colors[s] for s in df["Status"]])
plt.axhline(max_shortfall, linestyle="--", color="black", label="Re-evaluation threshold = 5 pts")
plt.axhline(0, color="gray", linewidth=0.8)
plt.title("Red Investments Fail Both Criteria and Need Re-evaluation", fontweight="bold", color="#1e3a8a")
plt.xlabel("Investment")
plt.ylabel("Shortfall vs. Expected ROI (pts)")
plt.legend()
plt.savefig("investment_review_solution/scene3_evidence.png", bbox_inches="tight")
plt.show()

to_review = df[df["Status"] == "Re-evaluate"]
at_risk = to_review["Capital_SAR_M"].sum()
share = at_risk / df["Capital_SAR_M"].sum() * 100

print("\n--- RE-EVALUATION LIST ---")
print(to_review[["Investment", "Capital_SAR_M", "Expected_ROI", "Actual_ROI", "Shortfall"]].to_string(index=False))
print(f"Capital at risk: SAR {at_risk:.1f}M ({share:.0f}% of portfolio)")

# ==========================================================
# Task 5 — BLUF Memo
# ==========================================================
bluf = {
    "Bottom Line": "Re-evaluate " + ", ".join(to_review["Investment"]) + ".",
    "Why": f"All return below the {hurdle_rate}% hurdle and trail expectations by {max_shortfall}+ points, tying up SAR {at_risk:.1f}M.",
    "Ask": "Approve a 90-day re-evaluation with a continue / restructure / exit decision for each."
}

print("\n--- EXECUTIVE BLUF MEMO ---")
for key, value in bluf.items():
    print(f"{key}: {value}")

# ==========================================================
# Task 6 & 7 — Self-Edit & Export
# ==========================================================
df.to_csv("investment_review_solution/portfolio_review.csv", index=False)
print("\n[SUCCESS] Saved project files into 'investment_review_solution' directory.")
```
